# 05 — Subir una submission a Kaggle

Utilidad para subir un CSV de `submission/` con el comando de Kaggle CLI, sin tener que escribirlo a mano cada vez.

**No es parte de la entrega** (no es `train.ipynb` ni `predict.ipynb`), es solo una herramienta.

**Esta celda la corro a mano.** Elijo el archivo y el mensaje en la celda de abajo, y recién cuando la corro se sube de verdad a Kaggle — consume una de las 5 submissions diarias.

Máximo 5 por día, al final se eligen 2.

```python
import sys
sys.path.append("..")

from pathlib import Path
from datetime import datetime

carpeta = Path("../submission")
candidatos = sorted(carpeta.glob("submission_*.csv"), key=lambda p: p.stat().st_mtime, reverse=True)

print(f"{'archivo':35} {'modificado':20} {'tamaño (KB)'}")
for archivo in candidatos:
    mod = datetime.fromtimestamp(archivo.stat().st_mtime).strftime("%Y-%m-%d %H:%M")
    kb = archivo.stat().st_size // 1024
    print(f"{archivo.name:35} {mod:20} {kb}")
```

## Elegir y subir

Antes de correr esta celda, chequeo que el archivo elegido tenga la fecha de modificación esperada (que sea de la última corrida completa de `train.ipynb` + `predict.ipynb`, no de una corrida vieja o mezclada — ver la lista de arriba).

Edito `archivo` y `mensaje` con lo que corresponda, y recién ahí corro la celda.

```python
import subprocess

archivo = "submission_beto_entrega.csv"  # <-- cambiá esto por el candidato que quieras subir
mensaje =  "P7c + duracion + texto (logistica + BETO)/2 (CV 0.9435, solo semilla 42)"  # <-- descripción corta del modelo

ruta = carpeta / archivo
assert ruta.exists(), f"No existe {ruta}"

subprocess.run(
    ["kaggle", "competitions", "submit", "-c", "ap-2026-q-2", "-f", str(ruta), "-m", mensaje],
    check=True,
)
```
